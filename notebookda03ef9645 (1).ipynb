```python

import os
import json
import subprocess
from pathlib import Path

# वीडियो ढूँढना
extensions = [".mp4", ".mov", ".avi", ".mkv"]
videos = []

for folder in [Path("/kaggle/input"), Path("/kaggle/working")]:
    if folder.exists():
        for f in folder.rglob("*"):
            if f.is_file() and f.suffix.lower() in extensions:
                videos.append(f)

print("Videos Found:", len(videos))

for i, f in enumerate(videos):
    print(i, f)

if not videos:
    print("ERROR: Video not found. Check Kaggle Add Data.")
else:
    VIDEO_PATH = str(videos[0])

    result = subprocess.run(
        ["ffprobe", "-v", "error",
         "-show_format", "-show_streams",
         "-of", "json", VIDEO_PATH],
        capture_output=True,
        text=True
    )

    if result.returncode != 0:
        print("FFprobe Error:", result.stderr)
    else:
        info = json.loads(result.stdout)
        fmt = info.get("format", {})

        print("\nVIDEO INFORMATION")
        print("File:", VIDEO_PATH)
        print("Duration:", fmt.get("duration", "Unknown"), "seconds")
        print("Size:", round(int(fmt.get("size", 0)) / 1048576, 2), "MB")

        for stream in info.get("streams", []):
            if stream.get("codec_type") == "video":
                print("Resolution:", stream["width"], "x", stream["height"])
                print("FPS:", stream.get("r_frame_rate"))

        print("\nVIDEO SUCCESSFULLY READ!")
```

```python

from scenedetect import open_video, SceneManager
from scenedetect.detectors import ContentDetector

# तुम्हारा वीडियो
video = open_video(VIDEO_PATH)

# Scene Detection
scene_manager = SceneManager()
scene_manager.add_detector(
    ContentDetector(threshold=27.0)
)

scene_manager.detect_scenes(video)

scenes = scene_manager.get_scene_list()

print("TOTAL SCENES:", len(scenes))

for i, scene in enumerate(scenes):
    start = scene[0].get_seconds()
    end = scene[1].get_seconds()

    print(
        f"Scene {i+1}: "
        f"{start:.2f}s to {end:.2f}s"
    )
```

```python

import subprocess
import numpy as np
from faster_whisper import WhisperModel

# Step 1: Video se audio extract karo
print("1. Extracting Audio...")

result = subprocess.run(
    [
        "ffmpeg", "-nostdin", "-v", "error",
        "-i", VIDEO_PATH,
        "-map", "0:a:0",
        "-ac", "1",
        "-ar", "16000",
        "-f", "f32le",
        "-"
    ],
    capture_output=True
)

if result.returncode != 0:
    raise RuntimeError(
        "Audio extraction failed: " +
        result.stderr.decode(errors="replace")
    )

audio = np.frombuffer(result.stdout, dtype=np.float32)

if len(audio) == 0:
    raise ValueError("Video mein audio nahi mila!")

print("Audio extracted successfully!")

# Step 2: Load Whisper AI
print("2. Loading Whisper AI...")

model = WhisperModel(
    "base",
    device="cuda",
    compute_type="float16"
)

# Step 3: Speech recognition
print("3. AI is understanding your video...")

segments, info = model.transcribe(
    audio,
    beam_size=5
)

# Step 4: Print subtitles
print("\nDetected Language:", info.language)
print("\nVIDEO TRANSCRIPT:")

transcript = list(segments)

for s in transcript:
    print(
        f"{s.start:.2f}s - {s.end:.2f}s: {s.text}"
    )

print("\nWHISPER TASK COMPLETED!")
```

```python

import os, subprocess, wave
import numpy as np

# =====================================
# 1. EXISTING KAGGLE FILES
# =====================================
VIDEO = "/kaggle/input/datasets/anujkumar0080008/aiveo8/1119.mp4"
VOICE = "/kaggle/input/datasets/anujkumar0080008/voice-data/voice.mp3"

OUT = "/kaggle/working/FINAL_AI_VIDEO.mp4"
SRT = "/kaggle/working/captions.srt"
BGM = "/kaggle/working/ai_background.wav"
SFX = "/kaggle/working/ai_effects.wav"

assert os.path.exists(VIDEO), "Original video nahi mila!"
assert os.path.exists(VOICE), "Voice file nahi mili!"

# =====================================
# 2. GET DURATIONS
# =====================================
def duration(path):
    return float(subprocess.check_output([
        "ffprobe", "-v", "error",
        "-show_entries", "format=duration",
        "-of", "default=noprint_wrappers=1:nokey=1",
        path
    ]).decode().strip())

vd = duration(VIDEO)
ad = duration(VOICE)
sr = 44100
n = int(ad * sr)
t = np.arange(n) / sr

print("Video:", vd, "seconds")
print("Voice:", ad, "seconds")

# =====================================
# 3. CREATE ORIGINAL BACKGROUND MUSIC
# =====================================
# Soft suspense/explainer ambient music
slow = 0.5 + 0.5 * np.sin(2*np.pi*0.10*t)
pad = (
    np.sin(2*np.pi*110*t) * 0.35 +
    np.sin(2*np.pi*164.81*t) * 0.20 +
    np.sin(2*np.pi*220*t) * 0.12
) * (0.35 + 0.65*slow)

# Gentle fade in/out
fade = np.minimum(1, t/2) * np.minimum(1, (ad-t)/2)
music = pad * fade * 0.10

stereo_music = np.column_stack([music, music])
with wave.open(BGM, "wb") as w:
    w.setnchannels(2)
    w.setsampwidth(2)
    w.setframerate(sr)
    w.writeframes((np.clip(stereo_music, -1, 1)*32767).astype("<i2").tobytes())

# =====================================
# 4. CREATE TIMED SOUND EFFECTS
# =====================================
effects = np.zeros(n, dtype=np.float32)

def add_coin(start, freq=1100, length=0.65, volume=0.20):
    i = int(start*sr)
    count = min(int(length*sr), n-i)
    if count <= 0:
        return
    x = np.arange(count)/sr
    sound = (
        np.sin(2*np.pi*freq*x) +
        0.45*np.sin(2*np.pi*freq*1.51*x)
    ) * np.exp(-7*x) * volume
    effects[i:i+count] += sound.astype(np.float32)

# Effects timed to the narration
add_coin(8.7, 1250)
add_coin(20.4, 900)
add_coin(22.5, 650)
add_coin(27.4, 1400)
add_coin(30.5, 750)

stereo_sfx = np.column_stack([effects, effects])
with wave.open(SFX, "wb") as w:
    w.setnchannels(2)
    w.setsampwidth(2)
    w.setframerate(sr)
    w.writeframes((np.clip(stereo_sfx, -1, 1)*32767).astype("<i2").tobytes())

# =====================================
# 5. HINGLISH CAPTIONS
# =====================================
captions = """1
00:00:04,210 --> 00:00:06,210
Agar aap 10 rupaye ki...

2
00:00:06,210 --> 00:00:10,210
Agar aap 10 rupaye ke coin mein drill se ched karke dhaaga baandh do

3
00:00:10,210 --> 00:00:13,210
Aur use vending machine mein daalkar wapas kheecho, to kya free mein samaan milega?

4
00:00:13,210 --> 00:00:15,210
Lekin aisa bilkul nahi hoga.

5
00:00:15,210 --> 00:00:17,210
Aapko lagega ki item selector ke button ko kheenchne se

6
00:00:17,210 --> 00:00:20,210
Maggi bhi mil jaayegi aur coin bhi wapas aa jaayega.

7
00:00:20,210 --> 00:00:22,210
Lekin jaise hi coin andar jaata hai,

8
00:00:22,210 --> 00:00:26,210
Uske baad ek optical laser sensor uska size aur weight verify karta hai.

9
00:00:27,210 --> 00:00:30,210
Aur coin aage badhte hi ye one-way metal lock ban jaata hai.

10
00:00:30,210 --> 00:00:32,210
To coin ko upar kheenchne par wahi phansa leta hai.

11
00:00:32,210 --> 00:00:34,180
Aur aapki chori pakdi jaati hai.
"""

with open(SRT, "w", encoding="utf-8") as f:
    f.write(captions)

# =====================================
# 6. COMPLETE VIDEO EDIT
# =====================================
# Stretch video gently to match full voice duration.
# This avoids cutting the voice or adding a black screen.
speed_factor = ad / vd

video_filter = (
    f"setpts=PTS*{speed_factor},"
    "scale=1080:1920:force_original_aspect_ratio=decrease,"
    "pad=1080:1920:(ow-iw)/2:(oh-ih)/2,"
    f"subtitles={SRT}:force_style='FontName=DejaVu Sans,"
    "FontSize=20,PrimaryColour=&H00FFFFFF,"
    "OutlineColour=&H00000000,BorderStyle=1,"
    "Outline=2,Shadow=1,Alignment=2,MarginV=130'"
)

cmd = [
    "ffmpeg", "-y",
    "-i", VIDEO,
    "-i", VOICE,
    "-i", BGM,
    "-i", SFX,
    "-filter_complex",
    "[1:a]aresample=44100,volume=1.0[v];"
    "[2:a]volume=0.18[m];"
    "[3:a]volume=0.35[s];"
    "[v][m][s]amix=inputs=3:duration=first:dropout_transition=2,"
    "alimiter=limit=0.95[a]",
    "-vf", video_filter,
    "-map", "0:v:0",
    "-map", "[a]",
    "-t", str(ad),
    "-c:v", "libx264",
    "-preset", "veryfast",
    "-crf", "20",
    "-c:a", "aac",
    "-b:a", "192k",
    "-movflags", "+faststart",
    OUT
]

subprocess.run(cmd, check=True)

print("\nSUCCESS! FINAL VIDEO READY")
print("File:", OUT)
print("Exists:", os.path.exists(OUT))
print("Duration:", duration(OUT))
```
